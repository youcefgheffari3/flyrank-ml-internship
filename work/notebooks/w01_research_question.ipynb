## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research question

**Can content and search-performance signals help prioritize which pages should be reviewed for a potential refresh, so that limited content-maintenance resources are directed toward the pages with the greatest potential opportunity?**

### The decision

The decision is which pages a content team should review first, given limited time and resources. The analysis will investigate whether available signals, such as changes in search impressions, clicks, click-through rate (CTR), and other relevant content-performance indicators, can help distinguish pages that may deserve attention from those that can continue to be monitored.

### The unit of analysis

The primary unit of analysis will be an individual content page. Where the dataset provides daily performance observations, I will also examine page-level performance over time to identify changes and trends. The final unit of analysis will depend on the available fields and the coverage of the starter dataset.

### The output

The intended output is a ranked list of content pages, accompanied by a transparent score or explanation of why each page has been prioritized. Initially, I will establish a simple baseline using observable performance signals. If the data supports it, I will compare this baseline with a machine learning approach to determine whether the model provides additional value.

### The action someone could take

A content editor or SEO specialist could use the ranking to decide which pages to investigate first. Depending on the evidence and subsequent human review, a page might be considered for refreshing, expanding, protecting, pruning, or continued monitoring. The score would support the decision rather than automatically determine the action.

### The cost of a wrong recommendation

A false positive could cause the team to spend time and resources updating a page that did not need attention, delaying work on more valuable pages. A false negative could cause the team to overlook a page with a genuine opportunity for improvement, potentially allowing its performance to deteriorate further. An inappropriate recommendation to prune or substantially rewrite a page could also damage useful content or organic search traffic.

For this reason, the ranking should be evaluated not only by its predictive performance, but also by the usefulness and potential cost of the decisions it supports.

### Why data and machine learning may help

The dataset may contain multiple performance signals and observations over time that are difficult to assess consistently by manually inspecting every page. Data analysis can help identify patterns and establish transparent prioritization rules. Machine learning may be useful if it can identify meaningful patterns beyond those captured by a simple baseline and if the available data supports a valid evaluation.

However, a model score alone cannot establish that refreshing a page will improve its performance. Search performance can be affected by factors outside the dataset, and observational patterns do not prove causation. I will therefore treat the ranking as a decision-support tool and evaluate the strength and limitations of the evidence before making recommendations.

The initial analysis will focus on the following questions:

- How many usable content records are available, and how many distinct clients do they represent?
- How many pages show a declining performance trend compared with those showing other trends?
- Which observable signals appear potentially useful for prioritizing pages for human review?

I will support these questions with executable code cells that calculate the relevant statistics directly from the starter CSV. I will treat these results as descriptive evidence, not proof that refreshing a page will improve its future performance.

```python

import pandas as pd

# Load the anonymized FlyRank starter dataset
url = "https://raw.githubusercontent.com/youcefgheffari3/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python

# Section 3: Quick look at the FlyRank starter dataset

n = len(df)

# 1. Pages currently labelled as declining
declining = df["trend_direction"].eq("down")

# 2. Pages with at least 500 impressions in the last 90 days
visible = df["impressions_90d"].ge(500)

# 3. Pages that meet both conditions
declining_visible = declining & visible

print(f"Total content records: {n:,}")

print(
    f"\nRecords labelled as declining: "
    f"{declining.sum():,} ({declining.mean() * 100:.1f}%)"
)

print(
    f"Records with at least 500 impressions: "
    f"{visible.sum():,} ({visible.mean() * 100:.1f}%)"
)

print(
    f"Records both declining and with at least 500 impressions: "
    f"{declining_visible.sum():,} "
    f"({declining_visible.mean() * 100:.1f}%)"
)
```
